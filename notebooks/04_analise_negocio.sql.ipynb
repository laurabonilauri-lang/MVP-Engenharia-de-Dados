```python
%sql

-- 1. Qual é o tempo médio de entrega e a taxa de atraso por estado (UF)?

SELECT 
    c.customer_state AS uf,
    COUNT(f.order_id) AS total_pedidos,
    ROUND(AVG(f.delivery_time_days), 1) AS tempo_medio_entrega_dias,
    ROUND((SUM(f.is_delayed) / COUNT(f.order_id)) * 100, 2) AS percentual_atraso
FROM workspace.default.fact_orders f
JOIN workspace.default.dim_customers c ON f.customer_id = c.customer_id
GROUP BY c.customer_state
ORDER BY tempo_medio_entrega_dias DESC;
```

```python
%sql
-- 2. Quais são as 10 categorias de produtos mais vendidas e o faturamento total?
SELECT 
    p.product_category_name AS categoria,
    COUNT(DISTINCT f.order_id) AS total_pedidos,
    SUM(f.total_items) AS total_itens_vendidos,
    ROUND(SUM(f.total_products_value), 2) AS faturamento_total
FROM workspace.default.fact_orders f
JOIN workspace.default.silver_order_items i ON f.order_id = i.order_id
JOIN workspace.default.dim_products p ON i.product_id = p.product_id
GROUP BY p.product_category_name
ORDER BY faturamento_total DESC
LIMIT 10;
```

```python
%sql
-- 3. Qual é a distribuição dos métodos de pagamento e o valor médio gasto?
SELECT 
    p.payment_type AS metodo_pagamento,
    COUNT(DISTINCT p.order_id) AS total_pedidos,
    ROUND(SUM(p.payment_value), 2) AS receita_total,
    ROUND(AVG(p.payment_value), 2) AS ticket_medio
FROM workspace.default.silver_payments p
JOIN workspace.default.fact_orders f ON p.order_id = f.order_id
GROUP BY p.payment_type
ORDER BY receita_total DESC;
```
